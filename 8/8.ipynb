```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.optimizers import SGD
from tensorflow.keras.preprocessing.image import ImageDataGenerator

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
# !wget https://github.com/SVizor42/ML_Zoomcamp/releases/download/straight-curly-data/data.zip
# !unzip data.zip
```

```python
model = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(200, 200, 3)),
    MaxPooling2D(pool_size=(2, 2)),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

### Q1. The best loss function for a binary classification problem:

#### binary crossentropy

```python
model.compile(
    optimizer=SGD(learning_rate=0.002, momentum=0.8),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

### Q2. The total number of parameters of the model:

```python
model.summary()
```

```python
image_datagen = ImageDataGenerator(rescale=1./255)

train_gen = image_datagen.flow_from_directory(
    directory='data/train/',
    target_size=(200, 200),
    batch_size=20,
    shuffle=True,
    class_mode='binary',
)

test_gen = image_datagen.flow_from_directory(
    directory='data/test/',
    target_size=(200, 200),
    batch_size=20,
    shuffle=True,
    class_mode='binary',
)
```

```python
history = model.fit(
    train_gen,
    epochs=10,
    validation_data=test_gen
)
```

### Q3. The median of training accuracy for all the epochs for this model:

```python
round(np.median(history.history['accuracy']), 3)
```

### Q4. The standard deviation of training loss for all the epochs for this model:

```python
round(np.std(history.history['loss']), 3)
```

```python
train_datagen_2 = ImageDataGenerator(
    rescale=1./255,
    rotation_range=50,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)

train_gen_2 = train_datagen_2.flow_from_directory(
    directory='data/train/',
    target_size=(200, 200),
    batch_size=20,
    shuffle=True,
    class_mode='binary',
)
```

```python
history2 = model.fit(
    train_gen_2,
    epochs=10,
    validation_data=test_gen
)
```

### Q5. The mean of test loss for all the epochs for the model trained with augmentations:

```python
round(np.mean(history2.history['loss']), 2)
```

### Q6. The average of test accuracy for the last 5 epochs (from 6 to 10) for the model trained with augmentations:

```python
round(np.mean(history2.history['accuracy'][5:10]), 2)
```
